```python
!pip install xgboost

# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.svm import SVR
from xgboost import XGBRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import os
os.makedirs("outputs", exist_ok=True)
```

```python
# Update path if needed
df = pd.read_csv("/kaggle/input/datasets/nikhileshiyer5/marketing-and-product-assignment/marketing_and_product_performance.csv")

print("Shape:", df.shape)
df.head()
```

```python
df.info()
df.describe()

# Check missing values
df.isnull().sum()
```

```python
target = "Units_Sold"

num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

# Impute missing values
imputer_num = SimpleImputer(strategy="mean")
df[num_cols] = imputer_num.fit_transform(df[num_cols])

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Encode categorical variables
le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col])
```

```python
X = df.drop(columns=[target])
y = df[target]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
def evaluate_model(name, model):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    
    return [name, mae, rmse, r2]
```

```python
# Without scaling
X_ns = df.drop(columns=[target])
X_train_ns, X_test_ns, y_train_ns, y_test_ns = train_test_split(
    X_ns, y, test_size=0.2, random_state=42
)

model = RandomForestRegressor()
model.fit(X_train_ns, y_train_ns)
pred_ns = model.predict(X_test_ns)

# With scaling
model.fit(X_train, y_train)
pred_s = model.predict(X_test)

df_rq3 = pd.DataFrame({
    "Method": ["No Scaling", "With Scaling"],
    "R2": [
        r2_score(y_test_ns, pred_ns),
        r2_score(y_test, pred_s)
    ]
})

df_rq3
```

```python
df_rq3.to_csv("outputs/RQ3_preprocessing.csv", index=False)

plt.plot(df_rq3["Method"], df_rq3["R2"], marker="o")
plt.title("RQ3: Preprocessing Impact")
plt.savefig("outputs/RQ3_plot.pdf")
plt.show()
```
